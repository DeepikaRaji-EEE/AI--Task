```python
import tkinter as tk
from tkinter import scrolledtext
from google import genai


# API KEY
API_KEY = "AQ.Ab8RN6L_Fp-tIhdxqZ1rWx7W5vBYySjLDwrdP9P8btP_tzJZrg"

client = genai.Client(api_key=AQ.Ab8RN6L_Fp-tIhdxqZ1rWx7W5vBYySjLDwrdP9P8btP_tzJZrg)

# ==========================
# DEVELOPER PROMPT
# ==========================

DEVELOPER_PROMPT = """
You are an AI Coding Assistant.

Rules:
1. Answer only coding questions.
2. Support Python, Java, C, C++, JavaScript, SQL, HTML and CSS.
3. Explain code simply.
4. Generate clean code.
5. Help fix coding errors.
6. If the question is not coding related, reply:
'Sorry, I can only answer programming questions.'
"""

# ==========================
# BOT RESPONSE
# ==========================

def get_bot_response(user_input):

    prompt = f"""
{DEVELOPER_PROMPT}

User:
{user_input}
"""

    try:
        response = client.models.generate_content(
            model="gemini-2.5-flash",
            contents=prompt
        )

        return response.text

    except Exception as e:
        return f"Error:\n{e}"

# ==========================
# SEND MESSAGE
# ==========================

def send_message():

    user_text = entry.get().strip()

    if user_text == "":
        return

    chat.insert(tk.END, "You : " + user_text + "\n\n", "user")

    reply = get_bot_response(user_text)

    chat.insert(tk.END, "Bot : " + reply + "\n\n", "bot")

    chat.see(tk.END)

    entry.delete(0, tk.END)

# ==========================
# TKINTER WINDOW
# ==========================

root = tk.Tk()
root.title("GenAI Coding Assistant")
root.geometry("700x700")

chat = scrolledtext.ScrolledText(
    root,
    font=("Arial",12),
    wrap=tk.WORD
)

chat.pack(fill=tk.BOTH, expand=True, padx=10, pady=10)

chat.tag_config("user", foreground="blue")
chat.tag_config("bot", foreground="green")

bottom = tk.Frame(root)
bottom.pack(fill=tk.X, padx=10, pady=10)

entry = tk.Entry(bottom, font=("Arial",12))
entry.pack(side=tk.LEFT, fill=tk.X, expand=True)

entry.bind("<Return>", lambda event: send_message())

button = tk.Button(
    bottom,
    text="Send",
    command=send_message,
    font=("Arial",12,"bold")
)

button.pack(side=tk.LEFT, padx=10)

root.mainloop()
```
